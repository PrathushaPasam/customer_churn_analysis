```python

custom_palette = {
    "No": "#F4A6B8",   # soft pink
    "Yes": "#7B2CBF"   # purple
}
plt.figure(figsize=(3,4))
gb = df.groupby("Churn").agg({'Churn':"count"})
plt.pie(
    gb["Churn"],
    labels=gb.index,
    autopct="%1.2f%%",
    colors=[custom_palette[label] for label in gb.index]
)
plt.title("perecentage of churned customers")
plt.show()
```

```python

custom_palette = {
    "No": "#F4A6B8",   # soft pink
    "Yes": "#7B2CBF"   # purple
}

plt.figure(figsize=(4,5))

ct = pd.crosstab(df['SeniorCitizen'], df['Churn'], normalize='index') * 100
ct = ct.reset_index()

ct_melt = ct.melt(id_vars='SeniorCitizen',
                  value_name='Percentage',
                  var_name='Churn')

ax = sns.barplot(
    data=ct_melt,
    x='SeniorCitizen',
    y='Percentage',
    hue='Churn',
    palette=custom_palette
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%")

plt.ylabel("Percentage")
plt.title("Churn Percentage by Senior Citizen")
plt.show()

```

#comapritive a greater percentage of people in senior citizen category have churned

```python

custom_palette = {
    "No": "#F4A6B8",     # soft pink
    "Yes": "#7B2CBF"     # purple
}
fig, axes = plt.subplots(1, 3, figsize=(18, 5),)
axes = axes.flatten()


nx=sns.countplot(data=df, x="PhoneService", hue="Churn", palette=custom_palette,ax=axes[0])
for container in nx.containers:
    nx.bar_label(container)
axes[0].set_title("PhoneService vs Churn")

nx=sns.countplot(data=df, x="MultipleLines", hue="Churn", palette=custom_palette,ax=axes[1])
for container in nx.containers:
    nx.bar_label(container)
axes[1].set_title("MultipleLines vs Churn")

nx=sns.countplot(data=df, x="InternetService", hue="Churn", palette=custom_palette, ax=axes[2])
for container in nx.containers:
    nx.bar_label(container)
axes[2].set_title("InternetService vs Churn")

plt.show()

```

```python
custom_palette = {
    "No": "#F4A6B8",     # soft pink
    "Yes": "#7B2CBF"     # purple
}

fig, axes = plt.subplots(3, 2, figsize=(18, 16))  # removed constrained_layout=True
axes = axes.flatten()

nx = sns.countplot(data=df, x="OnlineSecurity", hue="Churn",palette=custom_palette, ax=axes[0])
for container in nx.containers:
    nx.bar_label(container)
axes[0].set_title("OnlineSecurity vs Churn")

nx = sns.countplot(data=df, x="OnlineBackup", hue="Churn",palette=custom_palette, ax=axes[1])
for container in nx.containers:
    nx.bar_label(container)
axes[1].set_title("OnlineBackup vs Churn")

nx = sns.countplot(data=df, x="DeviceProtection", hue="Churn",palette=custom_palette, ax=axes[2])
for container in nx.containers:
    nx.bar_label(container)
axes[2].set_title("DeviceProtection vs Churn")

nx = sns.countplot(data=df, x="TechSupport", hue="Churn",palette=custom_palette, ax=axes[3])
for container in nx.containers:
    nx.bar_label(container)
axes[3].set_title("TechSupport vs Churn")

nx = sns.countplot(data=df, x="StreamingTV", hue="Churn",palette=custom_palette, ax=axes[4])
for container in nx.containers:
    nx.bar_label(container)
axes[4].set_title("StreamingTV vs Churn")

nx = sns.countplot(data=df, x="StreamingMovies", hue="Churn",palette=custom_palette, ax=axes[5])
for container in nx.containers:
    nx.bar_label(container)
axes[5].set_title("StreamingMovies vs Churn")
for ax in axes:
    ax.set_ylim(0, ax.get_ylim()[1] * 1.10)

# ✅ spacing controls (vertical + horizontal)
fig.subplots_adjust(hspace=0.8, wspace=0.35)

plt.show()
```

"""Customers who do not have support-related services such as Online Security and Tech Support exhibit noticeably higher churn levels. Backup and device protection services show similar trends, while streaming
services appear to have a weaker association with churn."""

```python
custom_palette = {
    "No": "#F4A6B8",   # soft pink
    "Yes": "#7B2CBF"   # purple
}

plt.figure(figsize=(6,5))

nx = sns.countplot(
    x="PaymentMethod",
    data=df,
    hue="Churn",
    palette=custom_palette
)

for container in nx.containers:
    nx.bar_label(container)
plt.xticks(rotation=30, ha='right')
plt.subplots_adjust(bottom=0.25)  # increase bottom margin
plt.tight_layout()


plt.show()
```

"""Customers paying through Electronic Check demonstrate noticeably higher churn compared to those using automatic payment methods. This pattern suggests that customers enrolled in automated payment systems may exhibit 
stronger commitment or longer-term engagement, contributing to reduced churn risk."""
